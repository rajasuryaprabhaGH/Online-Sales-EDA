```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
df = pd.read_excel(next(iter(uploaded)))
df.head()
```

```python
df.describe()
```

```python
category_revenue = df.groupby("Product Category")["Total Revenue"].sum()

category_revenue.plot(kind="bar", figsize=(8,5))
plt.title("Total Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
df["Date"] = pd.to_datetime(df["Date"])

monthly_revenue = df.groupby(df["Date"].dt.to_period("M"))["Total Revenue"].sum()

monthly_revenue.plot(kind="line", figsize=(10,5), marker="o")
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
numeric_df = df[["Units Sold", "Unit Price", "Total Revenue"]]

correlation = numeric_df.corr()

plt.figure(figsize=(7,5))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Between Sales Variables")
plt.show()
```

```python
df.groupby("Region")["Total Revenue"].sum().sort_values(ascending=False)
```

```python
df.groupby("Payment Method")["Total Revenue"].sum().sort_values(ascending=False)
```

```python
region_revenue = df.groupby("Region")["Total Revenue"].sum().sort_values(ascending=False)

region_revenue.plot(kind="bar", figsize=(8,5))
plt.title("Total Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
print("KEY INSIGHTS")
print("-" * 40)

print("1. Total Revenue:", df["Total Revenue"].sum())
print("2. Total Units Sold:", df["Units Sold"].sum())
print("3. Highest Revenue Category:",
      df.groupby("Product Category")["Total Revenue"].sum().idxmax())
print("4. Highest Revenue Region:",
      df.groupby("Region")["Total Revenue"].sum().idxmax())
print("5. Highest Revenue Payment Method:",
      df.groupby("Payment Method")["Total Revenue"].sum().idxmax())
```

# Exploratory Data Analysis – Online Sales

## 1. Introduction
This project performs Exploratory Data Analysis (EDA) on an online sales dataset to understand sales performance, revenue trends, product categories, regional performance, payment methods, and relationships between numerical variables.

## 2. Dataset Overview
The dataset contains information about online sales transactions, including:
- Transaction ID
- Date
- Product Category
- Product Name
- Units Sold
- Unit Price
- Total Revenue
- Region
- Payment Method

## 3. Statistical Analysis
Statistical summaries were generated to understand the distribution of numerical variables such as Units Sold, Unit Price, and Total Revenue.

## 4. Visualizations
The following visualizations were created:
- Total Revenue by Product Category
- Monthly Revenue Trend
- Correlation Heatmap
- Total Revenue by Region

## 5. Correlation Analysis
Correlation analysis was performed between Units Sold, Unit Price, and Total Revenue to identify relationships between the numerical variables.

## 6. Key Insights
- The highest-revenue product category was identified.
- The highest-revenue region was identified.
- The payment method associated with the highest revenue was identified.
- Monthly revenue trends were analyzed.
- Relationships between sales variables were examined using correlation analysis.

## 7. Conclusion
The EDA provided useful insights into online sales performance and helped identify important patterns, revenue trends, and factors associated with sales performance.
